```python
ax = df['Age'].value_counts().sort_index().plot(kind='bar', figsize=(20, 10))
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Number of People per Age')

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

```python
bins = list(range(18,70,6)) #18,#24,#30
df['Age_Bin_6yr'] = pd.cut(df['Age'], bins=bins, include_lowest=True)

df.groupby('Age_Bin_6yr')['Medical Cost'].describe()
```
