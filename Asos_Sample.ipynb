```python
df = df.dropna(how='all')
df.shape[0]
```

```python
# .copy() tells pandas explicitly: make df_products a fully independent dataframe,
# not just a "view" into df. This avoids the SettingWithCopyWarning we saw,
# and prevents confusing bugs later where editing df_products might accidentally affect df
df_products = df.drop_duplicates(subset='sku').copy()
df_products.shape[0]
```

```python
# Remove the word "Now" or "From" from the price text, leaving just the number as a string
price_cleaned = df_products['price'].str.replace(r'(Now|From)\s*', '', regex=True, case=False)

# Convert cleaned text into an actual number (float = decimal number)
df_products['price_clean'] = pd.to_numeric(price_cleaned, errors='coerce')

# Check: how many failed to convert now?
df_products['price_clean'].isna().sum()
```

```python
df_products['price'].head(10)
```

```python
pd.to_numeric(df_products['price'], errors='coerce').isna().sum()
```

```python
# Find the actual rows where price failed to convert to a number
# ~ means "NOT" - so this grabs rows where the value is NOT null after coercion attempt
bad_prices = df_products[pd.to_numeric(df_products['price'], errors='coerce').isna()]

# Show the unique bad values (not all 12,097 rows, just what TYPES of bad values exist)
bad_prices['price'].unique()[:20]  # first 20 unique bad values
```

```python
# Remove the original text-based 'price' column since we now have 'price_clean' as a proper number
# axis=1 means "drop a column" (axis=0 would mean "drop a row")
# inplace=True means "apply this change directly to df_products" instead of returning a new copy
df_products.drop('price', axis=1, inplace=True)

# Optional: rename price_clean back to 'price' now that it's the only price column
# This just makes future code cleaner to read/write (df_products['price'] instead of ['price_clean'])
df_products.rename(columns={'price_clean': 'price'}, inplace=True)

# Confirm the columns look right now
df_products.columns
```

```python
# Sort products by price, highest first, and grab the top 20 most expensive
# ascending=False means sort from highest to lowest
top_20_priciest = df_products.sort_values('price', ascending=False).head(20)

# Create a horizontal bar chart (easier to read long product names than vertical bars)
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))  # width, height in inches - taller since we have 20 items to fit

# barh = horizontal bar chart. x=name (categories), height=price (bar length)
plt.barh(top_20_priciest['name'], top_20_priciest['price'])

# invert_yaxis flips the order so the MOST expensive item appears at the TOP of the chart
# (by default, matplotlib would put it at the bottom)
plt.gca().invert_yaxis()

plt.xlabel('Price ($)')
plt.ylabel('Product')
plt.title('Top 20 Most Expensive Products')
plt.tight_layout()  # prevents long product names from getting cut off

plt.show()
```

```python
# Define common clothing types to search for within the product name
# This lets us build a REAL category column by checking which keyword appears in each name
clothing_types = ['dress', 'top', 'jumper', 'jacket', 'coat', 'shirt', 'jeans',
                   'skirt', 'trouser', 'shorts', 'blazer', 'vest', 'hoodie', 'jumpsuit' , 'footwear', 'beauty' , 'activewear']

# For each product name, find which clothing type word appears in it (case-insensitive)
# If none match, label it 'other'
def extract_category(name):
    name_lower = str(name).lower()
    for word in clothing_types:
        if word in name_lower:
            return word
    return 'other'

# .apply() runs the extract_category function on every single row's 'name' value
df_products['category_clean'] = df_products['name'].apply(extract_category)

# Check how many products fall into each new category, and how big is 'other'
df_products['category_clean'].value_counts()
```

```python
import matplotlib.pyplot as plt

# Group by our newly-extracted category, then calculate average price per group
avg_price_by_category = df_products.groupby('category_clean')['price'].mean()

# Sort from highest average price to lowest, so the chart reads most->least expensive
avg_price_by_category = avg_price_by_category.sort_values(ascending=False)

plt.figure(figsize=(10, 6))  # width, height in inches

# Vertical bar chart - now only ~14 bars instead of 29,000, so this will render instantly
plt.bar(avg_price_by_category.index, avg_price_by_category.values)

plt.xlabel('Category')
plt.ylabel('Average Price ($)')
plt.title('Average Price by Category')

# Rotate x-axis labels 45 degrees so labels don't overlap each other
plt.xticks(rotation=45, ha='right')

plt.tight_layout()  # prevents labels from getting cut off at the edges
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count how many products fall into each category
# .value_counts() automatically sorts from most to least common
category_counts = df_products['category_clean'].value_counts()

plt.figure(figsize=(10, 6))  # width, height in inches

# Vertical bar chart - category on x-axis, count of items on y-axis
plt.bar(category_counts.index, category_counts.values)

plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.title('Number of Products by Category')

# Rotate x-axis labels 45 degrees so they don't overlap each other
plt.xticks(rotation=45, ha='right')

plt.tight_layout()  # prevents labels from getting cut off at the edges
plt.show()
```
